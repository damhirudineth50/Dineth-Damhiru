```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
penguins = pd.read_csv('/content/penguins.csv')
print(penguins.head())
```

Once you run that code block in your Colab notebook, it will display a table showing the first five rows of the dataset.

```python
print(penguins.describe())
print(penguins.info())
print(penguins["species"].value_counts())
print(penguins.isna().sum())
```

Once you run this, you will see a text summary of the data. To answer your questions, we need to look closely at the output generated by the last two lines.

```python
sns.histplot(data=penguins, x="body_mass_g", bins=20)
plt.title("Distribution of Penguin Body Mass")
plt.show()
```

To figure out the shape of the distribution, you will need to look closely at the graph this code generates. Here is a quick guide to reading the shape:

Symmetric: The left and right sides of the peak look roughly like mirror images (like a bell curve).

Skewed right: The bulk of the data is on the left, and a long "tail" stretches out further to the right side of the graph.

Skewed left: The bulk of the data is on the right, and a long "tail" stretches out further to the left side.

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.title("Bill Length vs. Bill Depth by Species")
plt.show()
```

When you run this code, the hue="species" part tells Seaborn to assign a different color to each of the three penguin species

```python
sns.boxplot(data=penguins, x="species", y="body_mass_g")
plt.title("Body Mass by Species")
plt.show()
```

Median: Look at the main colored boxes for each species. The horizontal line running through the inside of each box represents the median body mass.

Outliers: Look at the lines extending out from the top and bottom of the boxes (these are called "whiskers"). Any individual dots floating completely outside of those whiskers are considered outliers.

```python
numeric_cols = penguins.select_dtypes(include="number")
corr = numeric_cols.corr()

sns.heatmap(corr, annot=True, cmap="Greys")
plt.title("Correlation Heatmap")
plt.show()
```

1 means a perfect positive correlation (as one goes up, the other goes up).

-1 means a perfect negative correlation (as one goes up, the other goes down).

0 means no correlation.

```python
sns.pairplot(penguins, hue="species")
plt.show()
```
